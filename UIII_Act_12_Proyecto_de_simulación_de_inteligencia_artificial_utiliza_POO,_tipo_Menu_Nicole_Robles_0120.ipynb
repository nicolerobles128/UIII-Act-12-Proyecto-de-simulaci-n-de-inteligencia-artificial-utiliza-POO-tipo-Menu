```python
# ==========================================
#   IA PARA ÁRBITROS - PROGRAMACIÓN POO
# ==========================================

class Arbitro:

    # Constructor
    def __init__(self, disciplina, records, patrocinio,
                 club, categoria, dieta):

        self.disciplina = disciplina
        self.records = records
        self.patrocinio = patrocinio
        self.club = club
        self.categoria = categoria
        self.dieta = dieta


# Lista principal
lista_arbitros = []


# ==========================================
# FUNCIONES DEL SISTEMA
# ==========================================

def agregar_arbitro():

    print("\n--- AGREGAR ÁRBITRO ---")

    disciplina = input("Disciplina deportiva: ")
    records = input("Récords personales: ")
    patrocinio = input("Patrocinios: ")
    club = input("Club deportivo: ")
    categoria = input("Categoría: ")
    dieta = input("Dieta nutricional: ")

    nuevo = Arbitro(disciplina, records, patrocinio,
                    club, categoria, dieta)

    lista_arbitros.append(nuevo)

    print("\n✅ Árbitro agregado correctamente")


def listar_arbitro():

    print("\n========= LISTA DE ÁRBITROS =========")

    if len(lista_arbitros) == 0:
        print("No hay árbitros registrados")
    else:

        print("{:<5} {:<15} {:<18} {:<15} {:<18} {:<15} {:<20}".format(
            "No", "Disciplina", "Récords", "Patrocinio",
            "Club", "Categoría", "Dieta"
        ))

        print("-" * 110)

        contador = 1

        for arbitro in lista_arbitros:

            print("{:<5} {:<15} {:<18} {:<15} {:<18} {:<15} {:<20}".format(
                contador,
                arbitro.disciplina,
                arbitro.records,
                arbitro.patrocinio,
                arbitro.club,
                arbitro.categoria,
                arbitro.dieta
            ))

            contador += 1


def actualizar_arbitro():

    print("\n--- ACTUALIZAR ÁRBITRO ---")

    listar_arbitro()

    if len(lista_arbitros) > 0:

        posicion = int(input("Número de árbitro a modificar: "))

        arbitro = lista_arbitros[posicion - 1]

        arbitro.disciplina = input("Nueva disciplina: ")
        arbitro.records = input("Nuevo récord: ")
        arbitro.patrocinio = input("Nuevo patrocinio: ")
        arbitro.club = input("Nuevo club: ")
        arbitro.categoria = input("Nueva categoría: ")
        arbitro.dieta = input("Nueva dieta: ")

        print("\n✅ Árbitro actualizado correctamente")


def borrar_arbitro():

    print("\n--- BORRAR ÁRBITRO ---")

    listar_arbitro()

    if len(lista_arbitros) > 0:

        posicion = int(input("Número de árbitro a borrar: "))

        lista_arbitros.pop(posicion - 1)

        print("\n✅ Árbitro eliminado correctamente")


# ==========================================
# MENÚ PRINCIPAL
# ==========================================

opcion = ""

while opcion != "5":

    print("\n=================================")
    print("    IA INTELIGENTE DE ÁRBITROS")
    print("=================================")
    print("1. Agregar árbitro")
    print("2. Listar árbitros")
    print("3. Actualizar árbitro")
    print("4. Borrar árbitro")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        agregar_arbitro()

    elif opcion == "2":
        listar_arbitro()

    elif opcion == "3":
        actualizar_arbitro()

    elif opcion == "4":
        borrar_arbitro()

    elif opcion == "5":
        print("\n👋 Gracias por usar la IA de Árbitros")
        print("Programa realizado por Nicole Robles NL = 50")

    else:
        print("\n❌ Opción inválida")
```

```python
# ==========================================
#   IA PARA ÁRBITROS - PROGRAMACIÓN POO
# ==========================================

print("========================================")
print("   IA INTELIGENTE PARA ÁRBITROS")
print("========================================")


class Arbitro:

    # Lista principal
    lista_arbitros = []

    # Constructor
    def __init__(self, disciplina, records,
                 patrocinio, club,
                 categoria, dieta):

        self.disciplina = disciplina
        self.records = records
        self.patrocinio = patrocinio
        self.club = club
        self.categoria = categoria
        self.dieta = dieta

    # ==========================================
    # AGREGAR ÁRBITRO
    # ==========================================

    @classmethod
    def agregar_arbitro(cls):

        print("\n--- AGREGAR ÁRBITRO ---")

        disciplina = input("Disciplina deportiva: ")
        records = input("Récords personales: ")
        patrocinio = input("Patrocinios: ")
        club = input("Club deportivo: ")
        categoria = input("Categoría: ")
        dieta = input("Dieta nutricional: ")

        nuevo = cls(disciplina, records,
                    patrocinio, club,
                    categoria, dieta)

        cls.lista_arbitros.append(nuevo)

        print("\n✅ Árbitro agregado correctamente")

    # ==========================================
    # LISTAR ÁRBITROS
    # ==========================================

    @classmethod
    def listar_arbitro(cls):

        print("\n============= LISTA DE ÁRBITROS =============")

        if len(cls.lista_arbitros) == 0:

            print("No hay árbitros registrados")

        else:

            print("{:<5} {:<15} {:<18} {:<15} {:<18} {:<15} {:<20}".format(
                "No", "Disciplina", "Récords",
                "Patrocinio", "Club",
                "Categoría", "Dieta"
            ))

            print("-" * 115)

            contador = 1

            for arbitro in cls.lista_arbitros:

                print("{:<5} {:<15} {:<18} {:<15} {:<18} {:<15} {:<20}".format(
                    contador,
                    arbitro.disciplina,
                    arbitro.records,
                    arbitro.patrocinio,
                    arbitro.club,
                    arbitro.categoria,
                    arbitro.dieta
                ))

                contador += 1

    # ==========================================
    # ACTUALIZAR ÁRBITRO
    # ==========================================

    @classmethod
    def actualizar_arbitro(cls):

        print("\n--- ACTUALIZAR ÁRBITRO ---")

        cls.listar_arbitro()

        if len(cls.lista_arbitros) > 0:

            posicion = int(input("\nNúmero de árbitro a modificar: "))

            arbitro = cls.lista_arbitros[posicion - 1]

            arbitro.disciplina = input("Nueva disciplina: ")
            arbitro.records = input("Nuevo récord: ")
            arbitro.patrocinio = input("Nuevo patrocinio: ")
            arbitro.club = input("Nuevo club: ")
            arbitro.categoria = input("Nueva categoría: ")
            arbitro.dieta = input("Nueva dieta: ")

            print("\n✅ Árbitro actualizado correctamente")

    # ==========================================
    # BORRAR ÁRBITRO
    # ==========================================

    @classmethod
    def borrar_arbitro(cls):

        print("\n--- BORRAR ÁRBITRO ---")

        cls.listar_arbitro()

        if len(cls.lista_arbitros) > 0:

            posicion = int(input("\nNúmero de árbitro a borrar: "))

            cls.lista_arbitros.pop(posicion - 1)

            print("\n✅ Árbitro eliminado correctamente")


# ==========================================
# MENÚ PRINCIPAL
# ==========================================

opcion = ""

while opcion != "5":

    print("\n=================================")
    print("         MENÚ PRINCIPAL")
    print("=================================")
    print("1. Agregar árbitro")
    print("2. Listar árbitros")
    print("3. Actualizar árbitro")
    print("4. Borrar árbitro")
    print("5. Salir")

    opcion = input("\nSelecciona una opción: ")

    if opcion == "1":

        Arbitro.agregar_arbitro()

    elif opcion == "2":

        Arbitro.listar_arbitro()

    elif opcion == "3":

        Arbitro.actualizar_arbitro()

    elif opcion == "4":

        Arbitro.borrar_arbitro()

    elif opcion == "5":

        print("\n👋 Gracias por usar la IA de Árbitros")
        print("Programa realizado por Nicole Robles NL = 50")

    else:

        print("\n❌ Opción inválida")
```
